```python
def parse_input(prompt):
    while True:
        try:
            values = list(map(float, input(prompt).strip().split()))
            if not values:
                raise ValueError("Input cannot be empty.")
            return values
        except ValueError:
            print("Invalid input. Please enter numeric values separated by spaces.")
```

```python
def print_summary(name, values):
    print(f"\n{name} values:")
    for val in values:
        print(val)
    total = sum(values)
    mean = total / len(values)
    squared_total = sum(x ** 2 for x in values)
    print(f"Summation of {name} = {total}")
    print(f"Mean of {name} = {mean}")
    print(f"Summation of {name}^2 = {squared_total}")
    return total, mean, squared_total
```

```python
def compute_regression(x_vals, y_vals):
    if len(x_vals) != len(y_vals):
        raise ValueError("X and Y must have the same number of values.")
    
    n = len(x_vals)
    xy_products = [x * y for x, y in zip(x_vals, y_vals)]
    xy_sum = sum(xy_products)
    
    x_sum = sum(x_vals)
    y_sum = sum(y_vals)
    x_mean = x_sum / n
    y_mean = y_sum / n
    x_sq_sum = sum(x ** 2 for x in x_vals)
    x_sum_sq = (x_sum ** 2) / n
    
    numerator = xy_sum - (x_sum * y_sum / n)
    denominator = x_sq_sum - x_sum_sq
    
    if denominator == 0:
        raise ZeroDivisionError("Cannot compute slope: denominator is zero.")

    b1 = numerator / denominator
    b0 = y_mean - b1 * x_mean
    return round(b0, 4), round(b1, 4)
```
